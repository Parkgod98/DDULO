```python
import json
import pandas as pd

# ../../data/tmap_puzzle_get-off.json 읽기
with open('../raw/tmap_puzzle_get-off.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
'''
df를 DataFrame으로 변환

DataFrame의 각 열은 다음과 같음:
- subwayLine : api_response의 contents의 subwayLine 값
- stationName: api_response의 contents의 stationName 값
- stationCode: api_response의 contents의 stationCode 값
- updnLine : api_response의 contents의 stat의 updnLine 값
- dow: api_response의 contents의 stat의 i번째의 data의 dow 값
- hh : api_response의 contents의 stat의 i번째의 data의 hh 값
- mm : api_response의 contents의 stat의 i번째의 data의 mm 값
- getOffCarRate : api_response의 contents의 stat의 i번째의 data의 getOffCarRate 값
- prevStationName : api_response의 contents의 stat의 prevStationName 값
- prevStationCode : api_response의 contents의 stat의 prevStationCode 값
- startStationName : api_response의 contents의 stat의 startStationName 값
- startStationCode : api_response의 contents의 stat의 startStationCode 값
- endStationName : api_response의 contents의 stat의 endStationName 값
- endStationCode : api_response의 contents의 stat의 endStationCode 값
- statStartDate : api_response의 contents의 statStartDate 값
- statEndDate : api_response의 contents의 statEndDate 값
'''


records = []

for entry in data:
    contents = entry['api_response']['contents']
    subway_line = contents['subwayLine']
    station_name = contents['stationName']
    station_code = contents['stationCode']
    stat_start_date = contents['statStartDate']
    stat_end_date = contents['statEndDate']
    
    for stat in contents['stat']:
        updn_line = stat['updnLine']
        prev_station_code = stat['prevStationCode']
        prev_station_name = stat['prevStationName']
        start_station_code = stat['startStationCode']
        start_station_name = stat['startStationName']
        end_station_code = stat['endStationCode']
        end_station_name = stat['endStationName']
        
        for data_point in stat['data']:
            record = {
                'subwayLine': subway_line,
                'stationName': station_name,
                'stationCode': station_code,
                'updnLine': updn_line,
                'dow': data_point['dow'],
                'hh': data_point['hh'],
                'mm': data_point['mm'],
                'getOffCarRate': data_point['getOffCarRate'],
                'prevStationName': prev_station_name,
                'prevStationCode': prev_station_code,
                'startStationName': start_station_name,
                'startStationCode': start_station_code,
                'endStationName': end_station_name,
                'endStationCode': end_station_code,
                'statStartDate': stat_start_date,
                'statEndDate': stat_end_date
            }
            records.append(record)
df = pd.DataFrame.from_records(records)
```

```python
# df csv로 저장
df.to_csv('../../data/tmap_puzzle_raw_get-off.csv', index=False, encoding='utf-8-sig')
df
```
