```python
'''
../data/past_average_congestion.csv
파일을 읽어와 dataframe으로 로드
'''
import pandas as pd

df = pd.read_csv('../../data/past_average_congestion.csv')
df
```

```python
'''
stationName, dow, hh, mm 순으로 정렬
'''
df = df.sort_values(by=['stationName', 'dow', 'hh', 'mm'])
df
```

```python
'''
stationName=역삼역, dow=FRI, hh=18, mm=10 인 데이터 추출
'''
df_yeoksam = df[(df['stationName'] == '역삼역') & (df['dow'] == 'FRI') & (df['hh'] == 18) & (df['mm'] == 10)]
df_yeoksam
```

```python
df_gangnam = df[(df['stationName'] == '강남역') & (df['dow'] == 'FRI') & (df['hh'] == 18) & (df['mm'] == 10)]
df_gangnam
```
